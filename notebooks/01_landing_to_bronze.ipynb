```python
import sys, os
sys.path.append(os.path.abspath(".."))

from spark_session import create_spark_session

# Cria a sessão Spark
spark = create_spark_session()

# Caminhos das camadas
landing_path = "../data/landing/vendas.csv"
bronze_path = "../data/bronze/vendas_delta"

# Leitura dos dados da camada Landing (CSV)
df = (
    spark.read.option("header", True)
    .option("inferSchema", True)
    .csv(landing_path)
)

# Exibe os dados lidos
df.show()

# Escreve em formato Delta na camada Bronze
df.write.format("delta").mode("overwrite").save(bronze_path)

print("✅ Dados movidos de Landing → Bronze (Delta Table)")
```
