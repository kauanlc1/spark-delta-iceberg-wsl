```python
import sys, os
sys.path.append(os.path.abspath(".."))

from spark_session import create_spark_session
from pyspark.sql import functions as F

spark = create_spark_session()

# Caminho da camada Silver
silver_path = "../data/silver/vendas_limpo"

# Leitura dos dados da camada Silver (Delta)
df_silver = spark.read.format("delta").load(silver_path)

# Exibe os dados carregados (camada Silver)
df_silver.show()

# Agregação: total de vendas por produto (quantidade * valor)
df_gold = df_silver.groupBy("produto").agg(
    F.sum(F.col("quantidade") * F.col("valor")).alias("total_vendas"),
    F.sum("quantidade").alias("total_quantidade")
)

# Caminho da camada Gold
gold_path = "../data/gold/vendas_agg"

# Escreve os dados agregados na camada Gold (formato Delta)
df_gold.write.format("delta").mode("overwrite").save(gold_path)

print("✅ Silver → Gold finalizado.")
```
