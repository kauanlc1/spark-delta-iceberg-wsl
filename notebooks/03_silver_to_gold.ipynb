```python
import sys, os
sys.path.append(os.path.abspath(".."))

from spark_session import create_spark_session
from pyspark.sql import functions as F

spark = create_spark_session()

df_silver = spark.table("silver_vendas_delta")

# Printo os dados carregados na silver
df_silver.show()

# Manipulo os dados
df_gold = df_silver.groupBy("produto").agg(F.sum("valor").alias("total_vendas"))

# Escrevo na gold (Formato p delta)
df_gold.write.format("delta").mode("overwrite").saveAsTable("gold_vendas_delta")


print("✅ Silver → Gold finalizado.")
```
