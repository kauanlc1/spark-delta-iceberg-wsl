```python
import sys, os
sys.path.append(os.path.abspath(".."))

from spark_session import create_spark_session

spark = create_spark_session()

# Leitura dos dados do Delta (Camada Bronze)
df_bronze = spark.table("bronze_vendas_delta")

# Exibe os dados carregados (da camada Bronze)
df_bronze.show()

# Aplica alguma transformação (se necessário)
df_silver = df_bronze.dropDuplicates().withColumn("valor", df_bronze["valor"].cast("double"))

# Escreve os dados transformados na camada Silver (formato Delta)
df_silver.write.format("delta").mode("overwrite").saveAsTable("silver_vendas_delta")

print("✅ Bronze → Silver finalizado.")
```
