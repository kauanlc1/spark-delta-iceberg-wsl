```python
import sys, os
sys.path.append(os.path.abspath(".."))

from spark_session import create_spark_session

spark = create_spark_session()

# Caminho da camada Bronze
bronze_path = "../data/bronze/vendas_delta"

# Leitura dos dados do Delta (Camada Bronze)
df_bronze = spark.read.format("delta").load(bronze_path)

# Exibe os dados carregados (da camada Bronze)
df_bronze.show()

# Aplica alguma transformação (se necessário)
df_silver = df_bronze.filter(df_bronze["quantidade"] > 0)  # Exemplo de transformação simples

# Caminho da camada Silver
silver_path = "../data/silver/vendas_limpo"

# Escreve os dados transformados na camada Silver (formato Delta)
df_silver.write.format("delta").mode("overwrite").save(silver_path)

print("✅ Bronze → Silver finalizado.")
```
